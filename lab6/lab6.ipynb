```python
import pymc as pm
import numpy as np
import arviz as az

data = np.array([56, 60, 58, 55, 57, 59, 61, 56, 58, 60])
#a
print("pt a")

x = np.mean(data)
print("x=", x)

with pm.Model() as model:
    m = pm.Normal("m", mu=x, sigma=10)
    sg = pm.HalfNormal("sg", sigma=10)
    likelihood = pm.Normal("X", mu=m, sigma=sg, observed=data)
    trace = pm.sample(2000, tune=1000, return_inferencedata=True)

#b
print("pt b")
hdi_m = az.hdi(trace.posterior["m"], hdi_prob=0.95)
hdi_sg = az.hdi(trace.posterior["sg"], hdi_prob=0.95)

print("m: HDI 95% = [", hdi_m.m.values[0], ",", hdi_m.m.values[1], "]")
print("sg: HDI 95% = [", hdi_sg.sg.values[0], ",", hdi_sg.sg.values[1], "]")

mean_m = trace.posterior["m"].mean().values
mean_sg = trace.posterior["sg"].mean().values

#c
print("pt c")
freq_mean = np.mean(data)
freq_std = np.std(data, ddof=1)

print("Media esantionului=", freq_mean)
print("Bayes Media posterioara m=", mean_m)

print("Deviatia standard a esantionului=", freq_std)
print("Bayes Media posterioara sg=", mean_sg)

diff_m = mean_m - freq_mean
diff_sg = mean_sg - freq_std

print("Delta m =", diff_m)
print("Delta sg =", diff_sg)

#d
print("pt d")
with pm.Model() as model_strong_prior:
    m_strong = pm.Normal("m", mu=50, sigma=1)
    sg_strong = pm.HalfNormal("sg", sigma=10)
    likelihood_strong = pm.Normal("X", mu=m_strong, sigma=sg_strong, observed=data)
    trace_strong = pm.sample(2000, tune=1000, return_inferencedata=True, random_seed=42)

mean_m_strong = trace_strong.posterior["m"].mean().values
mean_sg_strong = trace_strong.posterior["sg"].mean().values

hdi_m_strong = az.hdi(trace_strong.posterior["m"], hdi_prob=0.95)
hdi_sg_strong = az.hdi(trace_strong.posterior["sg"], hdi_prob=0.95)

print("prior puternic")
print("m:", mean_m_strong, ", HDI 95% = [", hdi_m_strong.m.values[0], ",", hdi_m_strong.m.values[1], "]")
print("sg:", mean_sg_strong, ", HDI 95% = [", hdi_sg_strong.sg.values[0], ",", hdi_sg_strong.sg.values[1], "]")

print("prior slab")
print("delta m =", mean_m_strong - mean_m)
print("delta sg =", mean_sg_strong - mean_sg)
```
